```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
pd.set_option('display.max_columns', None)
df = pd.read_csv('home-credit-default-risk/application_train.csv')
print(df.shape)
```

```python
df.head()
```

```python
target_counts = df["TARGET"].value_counts().sort_index()

target_df = pd.DataFrame({
    "count": target_counts,
    "pct": (target_counts / len(df)).map("{:.2%}".format),
})
display(target_df)
```

```python
print(df.dtypes.value_counts())
df.info(show_counts=True)
```

```python
df.columns.tolist()
```

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(include=['object']).columns.tolist()

# SK_ID_CURR is an ID, not a feature; TARGET is the label
for c in ['SK_ID_CURR', 'TARGET']:
    if c in num_cols:
        num_cols.remove(c)

print(f"Numerical: {len(num_cols)}")
print(f"Categorical: {len(cat_cols)}")
print("\nCategorical cols:", cat_cols)
```

```python
for c in cat_cols:
    print(f"{c}  ({df[c].nunique()} unique)")
    print(df[c].value_counts(dropna=False).head(8))
    print("-" * 50)
```

```python
df[num_cols].describe().T
```

```python
miss = df.isnull().sum().sort_values(ascending=False)
miss = miss[miss > 0]
print(f"{len(miss)} columns have missing values\n")

missing_df = pd.DataFrame({
    "n_missing": miss,
    "pct_missing": (miss / len(df)).map("{:.1%}".format),
})

with pd.option_context("display.max_rows", None):
    display(missing_df)
```

```python
# Binary flags stored as int (0/1) — NOT continuous numerics; treat separately
doc_flags = [c for c in df.columns if c.startswith('FLAG_DOCUMENT_')]        # 20 cols
other_flags = ['FLAG_MOBIL','FLAG_EMP_PHONE','FLAG_WORK_PHONE','FLAG_CONT_MOBILE',
               'FLAG_PHONE','FLAG_EMAIL']
region_flags = ['REG_REGION_NOT_LIVE_REGION','REG_REGION_NOT_WORK_REGION',
                'LIVE_REGION_NOT_WORK_REGION','REG_CITY_NOT_LIVE_CITY',
                'REG_CITY_NOT_WORK_CITY','LIVE_CITY_NOT_WORK_CITY']

# Building/housing block — the 47 _AVG/_MODE/_MEDI cols, mostly >50% missing
building = [c for c in df.columns if c.endswith(('_AVG','_MODE','_MEDI'))]

# Credit bureau enquiry counts
amt_req = [c for c in df.columns if c.startswith('AMT_REQ_CREDIT_BUREAU')]

print(f"Doc flags: {len(doc_flags)}")
print(f"Building cols: {len(building)}")
print(f"Bureau req cols: {len(amt_req)}")
```
